# ORION Quant — first experiment
This notebook runs a backtest on **synthetic demo prices**. These prices are made up, not real market data.

```python
# If running in Google Colab, clone the repository and install its package first.
!git clone https://github.com/zakir-si/ORION-Quant.git
%cd ORION-Quant
!pip install -e .
```

```python
from orion.data import make_demo_prices
from orion.engine import run_backtest
from orion.reporting import summarize
from orion.strategies import moving_average_signal

prices = make_demo_prices(days=500, seed=7)
signal = moving_average_signal(prices['Close'], fast=10, slow=40)
result = run_backtest(prices, signal, starting_cash=100_000, fee_rate=0.001, slippage_rate=0.0005)
summarize(result)
```
